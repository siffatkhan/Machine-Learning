## Exercise 1 — Load Iris Dataset

Load the Iris dataset using sklearn's datasets module. Assign the data to a variable named `iris_data` and the target to `iris_target`.

```python
from sklearn import datasets
from sklearn.datasets import load_iris
# YOUR CODE HERE. BY SIFFAT ULLAH BCS (B) 7TH SEMESTER
iris = load_iris()
iris_data = iris.data
iris_target = iris.target

print(iris_data)
print(iris_target)
```

## Exercise 2 — Split the Dataset

Split the Iris dataset into training and testing sets. Use a test size of 20% and random state of 42. Assign the results to `X_train`, `X_test`, `y_train`, and `y_test`.

```python
# YOUR CODE HERE
X_train, X_test, y_train, y_test = train_test_split(iris_data,iris_target, test_size=0.8, random_state=42)
```

## Exercise 3 — Scale the Features

Scale the features using StandardScaler. Fit the scaler on the training data and transform both the training and testing data. Assign the scaled data to `X_train_scaled` and `X_test_scaled`.

```python
scaler = StandardScaler()

# YOUR CODE HERE

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## Exercise 4 — Initialize the MLP Classifier

Initialize an MLPClassifier with 3 hidden layers of 10 neurons each. Use a random state of 42.

```python
mlp = None

# YOUR CODE HERE

from sklearn.neural_network import MLPClassifier
mlp = MLPClassifier(hidden_layer_sizes=(10,10,10), random_state=42)
```

## Exercise 5 — Train the MLP Classifier

Train the MLPClassifier on the scaled training data.

```python
# YOUR CODE HERE
mlp.fit(X_train_scaled, y_train)
```

## Exercise 6 — Make Predictions

Use the trained MLPClassifier to make predictions on the scaled testing data. Assign the predictions to `y_pred`.

```python
y_pred = None
# YOUR CODE HERE

y_pred = mlp.predict(X_test_scaled)
```

## Exercise 7 — Evaluate the Model

Evaluate the model by calculating the accuracy score of the predictions. Assign the score to `accuracy`.

```python
accuracy = None
# YOUR CODE HERE

accuracy = accuracy_score(y_test, y_pred)
print(accuracy)
```

## Exercise 8 — Interpret the Results

Based on the accuracy score, determine if the model's performance is satisfactory. Assign `True` to `is_satisfactory` if accuracy is above 0.9, otherwise assign `False`.

```python
is_satisfactory = None
# YOUR CODE HERE
is_satisfactory = accuracy > 0.9
is_satisfactory
```
