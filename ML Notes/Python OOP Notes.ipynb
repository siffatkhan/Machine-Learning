# Python OOP — Revision Notes

## 1. Classes & Objects

```
class Student:
    def __init__(self, name, age):     # constructor - runs automatically on creation
        self.name = name
        self.age = age

    def greet(self):
        return f"Hello {self.name}"

s1 = Student("Ali", 12)
print(s1.name, s1.age, s1.greet())

# self refers to the specific object (instance) being worked with
```

## 2. Class Variables vs Instance Variables

```
class Student:
    school = "IMSciences"       # class variable - SHARED by all instances

    def __init__(self, name):
        self.name = name         # instance variable - unique to each object

s1 = Student("Ali")
s2 = Student("Sifat")
print(s1.school, s2.school)      # both share the same class variable
```

## 3. Instance / Class / Static Methods

```
class Student:
    school = "Imsciences"

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def instance_method(self):          # works with ONE object's data
        return self.name

    @classmethod
    def school_name(cls, new_school):   # works with the CLASS itself
        cls.school = new_school

    @staticmethod
    def add(a, b):                      # doesn't touch self or cls at all,
        return a + b                    # just a utility function living in the class

Student.school_name("LUMS")
print(Student.school)

# Instance method (self) -> works on one object's data
# Class method (cls)     -> works on class-wide/shared data
# Static method          -> pure utility, no access to self or cls
```

## 4. Encapsulation

```
class Student:
    def __init__(self, name):
        self.name = name        # public - accessible anywhere
        self._name = name       # protected - convention only, "please don't touch"
        self.__name = name      # private - name-mangled to _Student__name

    def get_name(self):
        return self.__name

s = Student("Ali")
print(s.get_name())

# Private attributes aren't truly unreachable - Python "mangles" the name
# to _ClassName__attr, so it's a strong convention, not hard security.
```

## 5. Getters, Setters & Deleters (`@property`)

```
class Student:
    def __init__(self, marks):
        self._marks = marks

    @property
    def marks(self):                 # getter - access like an attribute: s.marks
        return self._marks

    @marks.setter
    def marks(self, value):          # setter - validates before assigning: s.marks = 90
        if value < 0:
            raise ValueError("Marks cannot be negative")
        self._marks = value

    @marks.deleter                    # deleter - del s.marks
    def marks(self):
        del self._marks

s = Student(80)
print(s.marks)      # getter
s.marks = 90         # setter
del s.marks           # deleter
```

## 6. Inheritance — Single

```
class Person:
    def __init__(self, name):
        self.name = name
    def speak(self):
        return "I am a person"

class Student(Person):
    def __init__(self, name, roll):
        super().__init__(name)       # calls the parent's __init__
        self.roll = roll
    def speak(self):                  # overrides parent's method
        return "I am a student"

s = Student("Sifat", 101)
print(s.name)
print(s.speak())
```

## 7. Inheritance — Multiple

```
class A:
    def show(self): return "Class A"
class B:
    def show(self): return "Class B"
class C(A, B):
    pass

c = C()
print(c.show())     # "Class A" - Python checks A before B (MRO)
print(C.mro())       # shows the exact lookup order Python follows
```

## 8. Inheritance — Multilevel & Hierarchical

```
# Multilevel: A -> B -> C (a chain)
class Animal: pass
class Mammal(Animal): pass
class Dog(Mammal): pass

# Hierarchical: one parent, multiple children
class Vehicle: pass
class Car(Vehicle): pass
class Bike(Vehicle): pass

# MRO (Method Resolution Order): the exact order Python searches classes
# when resolving a method/attribute, using the C3 linearization algorithm.
# ClassName.mro() shows the order directly.
```

## 9. Polymorphism

```
class Animal:
    def sound(self): return "Some sound"
class Dog(Animal):
    def sound(self): return "Bark"
class Cat(Animal):
    def sound(self): return "Meow"

animals = [Dog(), Cat()]
for animal in animals:
    print(animal.sound())     # same method call, different behavior per object
```

## 10. Duck Typing

```
class Duck:
    def sound(self): return "Quack"
class Person:
    def sound(self): return "I'm quacking!"

for thing in [Duck(), Person()]:
    print(thing.sound())      # works for both - no shared parent class needed

# "If it walks like a duck and quacks like a duck, it's a duck."
# Python doesn't check an object's type before calling a method - if it has
# the method, it works.
```

## 11. Operator Overloading

```
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y
    def __add__(self, other):              # defines behavior of Point + Point
        return Point(self.x + other.x, self.y + other.y)
    def __str__(self):                       # defines print(obj) / str(obj)
        return f"({self.x}, {self.y})"

p1, p2 = Point(1, 2), Point(3, 4)
print(p1 + p2)         # -> (4, 6)
```

## 12. Abstraction

```
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass          # no implementation here - subclasses must override this

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14 * self.r ** 2

# Shape()        -> TypeError, can't instantiate an abstract class directly
c = Circle(5)
print(c.area())   # works

# This guarantees every subclass provides its own area() - a contract
# enforced by Python itself, not just a comment/convention.
```

## 13. Dunder (Magic) Methods

```
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):           # readable string - used by print()
        return f"{self.title} ({self.pages}p)"

    def __repr__(self):           # unambiguous string - used in debugging/REPL
        return f"Book('{self.title}', {self.pages})"

    def __len__(self):            # enables len(book_obj)
        return self.pages

    def __eq__(self, other):      # enables book1 == book2
        return self.title == other.title

b1 = Book("Python 101", 200)
print(b1)
print(len(b1))
```

## 14. Composition vs Inheritance

```
class Engine:
    def start(self):
        return "Engine started"

class Car:
    def __init__(self):
        self.engine = Engine()      # Car HAS an Engine - composition
    def start(self):
        return self.engine.start()

car = Car()
print(car.start())

# Inheritance = "is-a" relationship (Student is a Person)
# Composition = "has-a" relationship (Car has an Engine)
# Rule of thumb: favor composition over inheritance - more flexible,
# avoids deep, fragile inheritance chains.
```

## 15. Full Real-World Example

```
class BankAccount:
    bank_name = "HBL"

    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance

    def deposit(self, amount):
        self._balance += amount
        return self._balance

    def withdraw(self, amount):
        if amount > self._balance:
            return "Insufficient balance"
        self._balance -= amount
        return self._balance

    @property
    def balance(self):
        return self._balance

    @classmethod
    def change_bank(cls, name):
        cls.bank_name = name

    @staticmethod
    def bank_rules():
        return "Minimum balance is 0"

acc = BankAccount("Sifat", 1000)
print(acc.deposit(500))
print(acc.withdraw(300))
print(acc.balance)
print(BankAccount.bank_rules())
BankAccount.change_bank("Allied")
print(BankAccount.bank_name)
```
