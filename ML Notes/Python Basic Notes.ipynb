## 1. Numbers & Operators

```
print(100 // 30)   # Floor division -> 3   (drops decimal, rounds toward -inf)
print(100 / 30)    # True division  -> 3.3333333333333335
print(100 % 30)    # Modulus (remainder) -> 10
print(2 ** 5)       # Exponent -> 32

# -7 // 2 -> -4, not -3 (floor division rounds toward negative infinity)
print(-7 // 2)

# Comparison operators: == != > < >= <=
# Logical operators: and, or, not
# Augmented assignment: x += 1, x -= 1, x *= 2, x //= 2
```

## 3. Strings

Strings are immutable — every "modifying" method returns a *new* string.

```
s = " Python "

s[0]              # first char
s[-1]             # last char
s.lower()         # lowercase
s.upper()         # uppercase
len(s)            # length
s.isalpha()       # all letters?
s.isdigit()       # all digits?
s.strip()         # removes leading/trailing whitespace
"***Py***".strip("*")     # strips given chars, not just whitespace
s.replace("Py", "My")     # replace substring (returns new string)
s.split(" ")               # split into a list
"k" in "khani"             # membership check -> True
```

### String Formatting

```
name, age = "Sifat", 21

# f-strings (modern, preferred)
print(f"My name is {name} and I am {age} years old")
print(f"{3.14159:.2f}")          # -> 3.14 (2 decimal places)

# .format()
print("My name is {} and I am {}".format(name, age))

# Other useful string methods
"hello world".title()      # -> "Hello World"
"Hello".startswith("He")   # -> True
"Hello".endswith("lo")     # -> True
"-".join(["a", "b", "c"])  # -> "a-b-c"
" ".join(["I", "love", "Python"])  # -> "I love Python"
"Hello"[::-1]               # reverse a string using slicing
```

## 4. Lists (ordered, mutable, allow duplicates)

```
arr = [1, 9, 8, 2, 10, 20]

arr[0]; arr[-1]              # indexing
arr[1] = 10                  # modify
arr.insert(1, 99)            # insert at index
arr.pop()                    # remove & return last item
arr.pop(0)                   # remove & return item at index
arr.remove(3)                # remove first occurrence of value 3
len(arr)

# Slicing
arr[1:5]      # index 1 to 4
arr[:3]       # first 3
arr[::2]      # every 2nd element
arr[::-1]     # reversed
arr[4:2:-1]   # reverse slice

3 in arr                     # membership check

for x in arr:                 # simple loop
    print(x)

for y in range(5, 16, 2):     # start, stop, step
    print(y)

for index, x in enumerate(arr):        # index + value
    print(index, x)

names = ["Ali", "Sifat", "Ahmed"]
for i, name in enumerate(names, start=1):   # custom start index
    print(i, name)

arr.append(50)       # add to end
arr.extend([60, 70])  # add multiple
arr.sort()            # sort in place (ascending)
arr.sort(reverse=True)
arr.reverse()          # reverse in place
arr.clear()            # empty the list
```

### List Comprehensions

```
squares = [x**2 for x in range(10)]              # [0,1,4,9,...]
evens   = [x for x in range(20) if x % 2 == 0]   # filter with condition
pairs   = [(x, y) for x in range(3) for y in range(2)]  # nested loop
labels  = ["even" if x % 2 == 0 else "odd" for x in range(5)]  # if/else
```

### Copy vs Reference

```
a = [1, 2, 3]
b = a            # b points to SAME list - changes to b affect a
b.append(4)
print(a)         # [1, 2, 3, 4]

c = a.copy()     # or a[:] or list(a) - creates an independent copy
c.append(5)
print(a)         # unaffected
```

## 5. Tuples (ordered, immutable)

```
tuple1 = (4, 5, 6)
a, b, c = tuple1              # unpacking
tuple2 = (5,)                  # single-element tuple needs the trailing comma

tuple1[0]; tuple1[-1]
for x in tuple1:
    print(x)

tuple1.count(5)                # occurrences of a value
tuple1.index(6)                # index of a value

lst = list(tuple1)             # tuple -> list
tup = tuple(lst)               # list -> tuple

# Tuples can be used as dictionary keys (lists cannot, since keys must be immutable)
```

## 6. Sets (unordered, mutable, unique values only)

```
set1 = {1, 2, 3, 4}
set1.add(4)             # no effect - 4 already exists
set1.remove(2)          # removes 2, raises error if absent
set1.discard(5)         # removes if present, no error if absent

s2 = {10, 20, 30, 4}
set1 | s2               # union
set1 & s2               # intersection
set1 - s2                # difference
set1 ^ s2                # symmetric difference (in one, not both)

set1.update([7, 8])     # add multiple
set1.clear()             # empty it
3 in set1                # membership check

# frozenset() creates an immutable set - can be nested inside another set or used as a dict key
fs = frozenset([1, 2, 3])
```

## 7. Dictionaries (key-value pairs, keys unique & immutable)

```
d = {"name": "Sifat", "age": 21, "city": "Lakki"}

d.get("name")            # safe access -> None if missing
d["city"]                # direct access -> KeyError if missing
d.get("job", "N/A")      # safe access with a default value

d["city"] = "Lahore"     # update
d["caste"] = "Marwat"    # add new key

for key in d:
    print(key, d[key])

for key, value in d.items():
    print(key, "-", value)

del d["age"]              # delete a key
removed = d.pop("city")   # delete and return value

"name" in d                # key existence check

d.keys(); d.values(); d.items()
d.update({"country": "Pakistan"})   # merge/add multiple
d.clear()
```

## 8. Conditionals & Loops

```
x = 5

if x > 0:
    print("positive")
elif x == 0:
    print("zero")
else:
    print("negative")

# Ternary (conditional expression)
result = "even" if x % 2 == 0 else "odd"

while x > 0:
    x -= 1

for i in range(5):
    if i == 3:
        break        # exits loop entirely
    if i == 1:
        continue     # skips to next iteration
else:
    print("loop finished without break")   # for/while can have an else clause
```

## 9. Functions

```
def sifat(name):
    return name

def add(a, b):
    return a + b

def greet(name="Guest"):          # default parameter
    return "Hello " + name

def calc(a, b):
    return a + b, a - b, a * b     # returns a tuple -> unpack it
sum_val, diff_val, mul_val = calc(10, 5)

square = lambda x: x * x           # anonymous one-line function
```

### `*args` and `**kwargs`

```
def total(*args):                 # collects extra positional args into a tuple
    return sum(args)
total(1, 2, 3, 4)                  # -> 10

def show(**kwargs):               # collects extra keyword args into a dict
    for k, v in kwargs.items():
        print(k, v)
show(name="Sifat", age=21)
```

### Variable Scope

```
x = 10               # global variable
def foo():
    x = 5             # local variable - shadows global inside foo
    print(x)          # 5
foo()
print(x)              # 10 - unaffected

def bar():
    global x
    x = 99             # explicitly modifies the global variable
bar()
print(x)              # 99
```

Docstrings: `def f():\n    """Explains what f does."""` — accessible via `f.__doc__`.

## 10. Common Built-in Functions

```
arr = [5, 3, 8, 1]

len(arr); sum(arr); min(arr); max(arr)
sorted(arr)                        # returns a NEW sorted list (doesn't mutate)
sorted(arr, reverse=True)
sorted(arr, key=lambda x: -x)      # sort by custom key

range(5, 16, 2)                    # start, stop, step
enumerate(arr, start=1)            # (index, value) pairs
zip([1,2,3], ["a","b","c"])        # pairs elements from multiple iterables
map(int, ["1","2","3"])            # apply function to every element
filter(lambda x: x > 5, arr)       # keep elements matching a condition

all([True, True, False])           # -> False (all must be True)
any([False, False, True])          # -> True (at least one True)
```

## 11. Exception Handling

```
try:
    x = int(input("Enter a number: "))
    result = 10 / x
except ValueError:
    print("That wasn't a valid number")
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("No errors - result is", result)   # runs only if try succeeded
finally:
    print("This always runs")                 # cleanup code, runs no matter what

# Raising your own errors
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")

# Custom exception class
class InsufficientBalanceError(Exception):
    pass
```

## 14. Practice Question

```
arr = []
for _ in range(5):
    arr.append(int(input("Enter a number: ")))

max_val = arr[0]
min_val = arr[0]
total = 0

for x in arr:
    if x > max_val:
        max_val = x
    if x < min_val:
        min_val = x
    total += x

avg_val = total / len(arr)

result = {
    "max": max_val,
    "min": min_val,
    "average": avg_val
}

print(result)
```
